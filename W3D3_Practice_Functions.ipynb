```python
# Your code here
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(is_even(4))
print(is_even(7))
```

### Stretch challenge — Guess helper
Write a function `check_guess(guess, secret)` that returns the string `"low"`, `"high"`, or `"correct"` by comparing `guess` to `secret`. This is the brain of tomorrow's-style game — test it with a few values!

```python
# Your code here
import random 
 
def check_guess(guess , secret):
    
    if guess > secret:
        return "guess lower"
    elif guess < secret:
        return "guess higher"
    else:
        return "correct"
    
secret = random.randint(1,51)
guess = None
attempts = 0

while True:
    guess = int(input("Enter a number"))

    print(check_guess(guess, secret))
    attempts += 1

    if check_guess(guess, secret) == "correct":
        print("Total attempts" , attempts)
        break
```
